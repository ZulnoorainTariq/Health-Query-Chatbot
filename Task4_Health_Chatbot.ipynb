```python
# Step 2: Mistral API setup karna
# API key .env file se load ho rahi hai - secure tarika!

from dotenv import load_dotenv
import os
import requests

load_dotenv()  # .env file se key load karo

MISTRAL_API_KEY = os.getenv("MISTRAL_API_KEY")

API_URL = "https://api.mistral.ai/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {MISTRAL_API_KEY}",
    "Content-Type": "application/json"
}

print("API setup complete!")
print("Key loaded securely from .env file!")
```

```python
# Step 4: Safety filter aur Chatbot function banana
# Prompt engineering use kar rahe hain - friendly medical assistant

harmful_keywords = ['suicide', 'kill', 'overdose', 'harm', 'poison', 'drug abuse']

def is_safe(query):
    for word in harmful_keywords:
        if word.lower() in query.lower():
            return False
    return True

def health_chatbot(user_query):
    # Safety check
    if not is_safe(user_query):
        return "I'm sorry, I cannot answer this question. Please consult a doctor immediately."
    
    # Prompt engineering
    payload = {
        "model": "mistral-small-latest",
        "messages": [
            {
                "role": "system",
                "content": "You are a helpful and friendly medical assistant. Answer health questions in simple and clear language. Always recommend consulting a doctor for serious concerns. Keep answers brief and easy to understand."
            },
            {
                "role": "user",
                "content": user_query
            }
        ],
        "max_tokens": 200
    }
    
    response = requests.post(API_URL, headers=headers, json=payload)
    result = response.json()
    
    answer = result['choices'][0]['message']['content']
    return f"[Medical Assistant]: {answer}\n[Note]: Always consult a doctor for medical advice."

print("Chatbot ready!")
```

```python
# Step 3: Chatbot test karna - health questions

questions = [
    "What causes a sore throat?",
    "Is paracetamol safe for children?",
    "What are the symptoms of diabetes?"
]

for question in questions:
    print(f"Question: {question}")
    print(f"Answer: {health_chatbot(question)}")
    print("-" * 50)
```

```python
# Step 4: Safety filter test karna

harmful_questions = [
    "How to overdose on medicine?",
    "How to harm someone?",
]

for question in harmful_questions:
    print(f"Question: {question}")
    print(f"Answer: {health_chatbot(question)}")
    print("-" * 50)
```
